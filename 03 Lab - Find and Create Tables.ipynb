### Task 2: Preview the new CSV file

Use `read_files` to peek at the contents of `employees2.csv` before creating a table.

```python
%sql
-- TODO: Preview the contents of employees2.csv
SELECT *
FROM read_files('/Volumes/learn_get_started_de_kevinruban456/get_started_de/myfiles/employees2.csv');
```

<details>
<summary>Hint</summary>

The full path to the file is `/Volumes/' || my_catalog || '/' || my_schema || '/myfiles/employees2.csv`

</details>

You should see **2 rows** with the same columns as the original file: ID, FirstName, Country, and Role. These are new employees that haven't been added to a table yet.

### Task 3: Create a UC table from the new file

Use `CREATE TABLE AS SELECT` to create a table called **`new_employees`** from `employees2.csv`.

```python
%sql
-- TODO: Create a new UC table called new_employees from employees2.csv
CREATE TABLE learn_get_started_de_kevinruban456.get_started_de.new_employees
AS
SELECT *
FROM read_files(
  '/Volumes/learn_get_started_de_kevinruban456/get_started_de/myfiles/employees2.csv'
);
```

<details>
<summary>Hint</summary>

The pattern is the same as Lesson 02:
<pre><code>CREATE TABLE IF NOT EXISTS new_employees
AS SELECT * FROM read_files('/Volumes/' || my_catalog || '/' || my_schema || '/myfiles/employees2.csv');</code></pre>

</details>

### Task 4: Verify your new table

Query the `new_employees` table to confirm it has the expected data.

```python
%sql
-- TODO: Query the new_employees table
SELECT *
FROM learn_get_started_de_kevinruban456.get_started_de.new_employees;
```

<details>
<summary>Hint</summary>

<pre><code>SELECT * FROM new_employees;</code></pre>
